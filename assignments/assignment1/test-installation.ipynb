## Step 1: Load the tools we need

Python comes with a large **standard library** — code that is already installed and ready to use. To use a part of it, you `import` it.

Here we import four of them: `sys` and `platform` know about your Python installation and your computer, `os` works with file paths, and `datetime` gives us the current date and time.

This cell produces no output. That is expected — it only makes these tools available to the cells below.

```python
import sys
import platform
import os
from datetime import datetime
```

## Step 2: Collect information about your setup

This cell asks Python a few questions about itself and about your computer, and stores each answer in a **variable** so that we can use it further down.

The important one is `python_path`. It is the location of the Python program that is running this notebook. If your installation is set up the way this course expects, that path contains the word `miniforge`.

This cell also produces no output — it collects the answers, and the next cell displays them.

```python
# Information about your computer
operating_system = f"{platform.system()} {platform.release()}"
computer_name = platform.node()

# Information about the Python that is running this notebook
python_version = platform.python_version()
python_path = sys.executable

# Does that path point at a Miniforge installation?
uses_miniforge = "miniforge" in python_path.lower()

# The date and time right now
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
```

## Step 3: Build and display the report

Now we put those answers together into a short report and print it to the screen.

The `if` statement at the top picks one of two messages, depending on whether Miniforge was found. You will learn about `if` statements in week 3 — for now, just run the cell.

**Read the output.** This is exactly what gets saved to your submission file in the next step.

```python
if uses_miniforge:
    miniforge_status = "yes"
else:
    miniforge_status = "NO - please show this report to us in class"

report = f"""TECH2 - Installation report
Generated: {timestamp}

Operating system : {operating_system}
Computer name    : {computer_name}
Python version   : {python_version}
Python location  : {python_path}
Miniforge in use : {miniforge_status}

This notebook ran from start to finish without errors.
"""

print(report)
```

## Step 4: Save the report as a text file

This cell writes the report to a file called `tech2-setup-report.txt`, in the same folder as this notebook.

`encoding="utf-8"` tells Python how to store the characters. It matters here because Norwegian letters like æ, ø and å often appear in file paths, and this is what makes sure they are saved correctly.

The cell then prints the **full path** to the file it just created. Use that path to find the file on your computer.

```python
file_name = "tech2-setup-report.txt"

with open(file_name, "w", encoding="utf-8") as f:
    f.write(report)

full_path = os.path.abspath(file_name)

print("Report saved.")
print()
print("Your file is here:")
print(full_path)
```
