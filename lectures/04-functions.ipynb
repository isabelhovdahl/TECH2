# 04 - Functions

Up until now, we have executed our programs as a series of instructions. However, most programs consist of distinct groups of instructions designed to perform a specific task, i.e., routines. In Python, routines are known as *functions*.

A function is a block of code that only runs when it is called. Creating functions offers several advantages:
- **Code reusability:** define the function once and use it whenever needed — don't repeat yourself (DRY)!
- **Organization:** break the program down into smaller, manageable units, which improves code organization.
- **Readability:** giving descriptive names to functions can make the code easier to navigate (for you and others!).
- **Abstraction:** you don't need to know how the sausage is made (i.e., the code) to use a function — only how to call it.

This notebook shows how we can create our own **self-defined functions** in Python.

## Creating functions

A function is a **named** group of instructions that accomplishes a **specific task**.

We have already used several of Python's built-in functions, e.g. `print`, `len`, and `type`. We have also used a function from a third-party package: `randint` from `random`.

A function typically has four characteristics:
1. Name
2. Task
3. Input (also known as parameters)
4. Output

> 📝 **Note:** Not every function needs input, and not every function produces an explicit output — we'll see examples of both later in this notebook.

For example, we have used `len` to calculate the length of sequences, e.g. strings and lists:

1. Name: `len`
2. Task: calculate the length of a sequence
3. Input: a sequence
4. Output: an integer

````python
len([1, 2, 3, 4, 5, 6, 7])
````

To create our own functions, we must define the function *header* and the function *body*:

- Function header:
    1. `def` keyword
    2. Function name
    3. Sequence of inputs enclosed in parentheses
    4. A colon
- Function body:
    1. Indented block of code
    2. `return` statement

```
def function_name(param1, param2, ...):

    <code block>

    return <output>
```

Let's write a program that calculates the average of three numbers.

````python
num1 = 10
num2 = 9
num3 = 4

avg = (num1 + num2 + num3) / 3

print(avg)
````

To increase the reusability of our program, we can place this code inside a function called `avg3`.

````python
def avg3(num1, num2, num3):

    avg = (num1 + num2 + num3) / 3

    return avg
````

> 📝 **Note:** A function must be *defined* (i.e., its code cell must have been run) before it can be used elsewhere in the program — the order cells are *run* in matters, not just the order they appear in the notebook.

Once a function is defined, we can use it as many times as we want, each time supplying the *function call* with new inputs.

````python
mean = avg3(num1=10, num2=9, num3=4)

print(mean)
````

````python
mean = avg3(num1=1.7, num2=9, num3=7)

print(mean)
````

To further increase the reusability of our program, we can instead create a function called `avg_lst` that calculates the average of a *list* of numbers. The function takes only one input (the list of numbers) and returns the average of that list.

````python
def avg_lst(number_lst):

    lst_sum = sum(number_lst)     # sum of the numbers
    lst_length = len(number_lst)  # length of the list
    avg = lst_sum / lst_length    # calculate the mean

    return avg
````

We can now use the function to calculate the average of a list of numbers, independent of the length of the list.

````python
mean = avg_lst(number_lst=[10, 9, 4, 5, 9, 10, 0])

print(mean)
````

````python
mean = avg_lst(number_lst=[1, 2, 3])

print(mean)
````

### Non-value-returning functions

We have so far looked at **value-returning functions**:
- They use the `return` statement to produce an output.
- The output can be stored by assigning it to a variable.

There are also **non-value-returning functions**:
- They have no `return` statement...
- ...but they have other desirable side effects (e.g., printing something).
- Their output is not stored in a variable name.

For example, let's modify `avg_lst` so that it *displays* the average instead of returning it, by substituting the `return` statement with a `print` statement.

````python
def avg_lst(number_lst):

    lst_sum = sum(number_lst)
    lst_length = len(number_lst)
    avg = lst_sum / lst_length

    print(f"The mean is {avg:.2f}")  # no return statement
````

With non-value-returning functions, we call the function without assigning it to a variable.

````python
avg_lst(number_lst=[10, 9, 4, 5, 9, 10, 0])
````

> 📝 **Note:** If you assign the result of a non-value-returning function call to a variable, that variable will simply store `None` — Python's default return value when a function has no `return` statement.

## Function input

To call a function, we must pass **arguments** to the function's **parameters**. This is known as *parameter passing*.

There are several ways to pass arguments to a function's parameters.

The function `division` below takes two numbers as parameters (`num1` and `num2`) and returns the result of dividing the first by the second.

````python
def division(num1, num2):
    result = num1 / num2
    return result
````

**1. Keyword arguments:**

We can call a function by directly passing arguments to the parameter names in the function call.

````python
division(num1=2, num2=-5)
````

Note that the order of the arguments does not matter when using the `parameter=argument` syntax.

````python
division(num2=-5, num1=2)
````

**2. Positional arguments:**

Alternatively, we don't have to specify the parameter names in the function call — but then the order of the arguments matters. By default, the first argument is passed to the first parameter, the second to the second, and so on.

````python
division(2, -5)
````

Swapping the order of positional arguments changes which parameter each value is assigned to — and can therefore change the result.

````python
division(-5, 2)
````

We can also store arguments in variables first, and then pass the variable names to the function call.

> ⚠️ **Warning:** When passing arguments positionally, it's the *position* that determines which parameter receives which value — not the variable's name. This can be confusing when a variable happens to share its name with a parameter.

````python
num1 = 5
num2 = 10

division(num1, num2)
````

````python
division(num2, num1)
````

Even though the second call passes `num2` first, its value (`10`) is assigned to the function's *first* parameter — because position, not name, determines the assignment.

By default, a function must be called with the same number of arguments as parameters.

````python
# TypeError
division(5)
````

### Default parameter values

We can give a parameter a default value when defining the function. This makes it *optional* to pass an argument for that parameter in the function call.

````python
def division(num1, num2=10):
    result = num1 / num2
    return result
````

````python
division(10)
````

````python
division(10, 5)
````

## Function output

The return value of a function can be stored in a variable, or used directly as part of a larger expression.

Recall that `division` has one required parameter, `num1`, and one optional parameter, `num2` (with a default value of 10).

````python
def division(num1, num2=10):
    result = num1 / num2
    return result
````

We can store the result of a function call in a variable, and use that variable in another function call.

````python
res1 = division(6, 2)
res2 = division(12, res1)

print(res2)
````

Alternatively, we can pass a function call directly as an argument to another function call.

````python
res3 = division(12, division(6, 2))

print(res3)
````

We can also use multiple function calls within a single expression, or directly inside a conditional.

````python
division(2, 4) * division(8, 3)
````

````python
a = 5
b = 3

if division(a, b) >= 1:
    print("Equal or larger than 1")
else:
    print("Smaller than 1")
````

A function can also have multiple `return` statements, to control which value is returned under which condition.

````python
def check_answer(answer):
    if answer in ("A", "B"):
        return True
    else:
        return False

valid_answer = check_answer("A")
#valid_answer = check_answer("C")

print(valid_answer)
````

Finally, a function can return *multiple* values at once.

````python
def calculations(num1, num2):

    res1 = num1 * num2
    res2 = num1 / num2

    return res1, res2  # return two values

results = calculations(10, 5)

results
````

> 📝 **Note:** Multiple return values are actually packed into a single *tuple* — recall from the first notebook that tuples are created with parentheses `()` and can hold multiple values.

We can *unpack* a tuple directly into separate variables.

````python
res1, res2 = calculations(10, 5)

print(res1)
print(res2)
````

## Implementing and testing functions

When designing functions, there are a few conventions to follow and common pitfalls to avoid.

**1. Ensure that the function handles all scenarios:**

Failing to do so can result in errors later in the program.

````python
def check_number(num1, num2):
    if num1 > num2:
        return 0
````

````python
res = check_number(20, 10)

print(res)
````

````python
res = check_number(10, 20)

print(res)
````

> 📝 **Note:** The default return value of a function in Python is `None`, unless an explicit return value is specified with the `return` statement.

````python
# TypeError (because "res" is None)
res * 100
````

**2. Local versus global variables:**

Variables created *inside* a function are *local* variables. They are not available outside the function, unless explicitly returned.

````python
def exp2(num):
    exponent = 2  # local variable
    return num**exponent
````

We cannot access a local variable outside of the function where it was created.

````python
# NameError
exponent
````

Local variables are actually great! 🎉 They let us reuse variable names in different functions without any name conflicts. For example, `exponent` is local to both `exp2` and `exp3` below, and can have a different value in each, even though they share the same name.

````python
def exp3(num):
    exponent = 3  # local variable
    return num**exponent

exp3(4)
````

The *scope* of a variable is the part of the program where the variable is "visible," i.e., where we can access its value. Variables defined outside of any function have **global** scope, and are called *global* variables.

````python
num = 10  # global variable
````

Functions can read global variables even if they are not passed in explicitly as an argument.

````python
n = 2  # global variable

def exp2(num):
    return num**n  # uses the global variable "n"

exp2(num=4)
````

> ⚠️ **Warning:** Relying on global variables inside a function is generally considered bad practice. It breaks the idea of a function as a self-contained "black box" that only depends on its inputs, which makes code harder to reuse and reason about — and easier to break by accident.
>
> Reading a global *constant* that never changes (e.g. a fixed conversion rate) is common and reasonably safe. The real risk is a function silently depending on a variable that could change elsewhere in the program, without that dependency being visible in the function's parameters.

Instead, all variables a function needs should either be defined *inside* the function...

````python
def exp2(num):
    exponent = 2  # local variable
    return num**exponent

exp2(num=4)
````

...or passed in explicitly as a parameter.

````python
n = 2  # global variable

def exp2(num, exponent):
    return num**exponent

exp2(4, n)  # global variable passed as an explicit argument
````

**3. Mutable arguments:**

Unlike numbers and strings, lists and dictionaries are *mutable*. If a function modifies a list or dictionary that was passed in as an argument, the change is visible outside the function too — even without a `return` statement.

````python
def add_item(lst):
    lst.append("new item")

fruits = ["apple", "banana"]
add_item(fruits)

print(fruits)
````

Compare this to passing a number, which is immutable — reassigning the parameter inside the function has no effect on the original variable.

````python
def add_one(num):
    num = num + 1

x = 5
add_one(x)

print(x)
````

> ⚠️ **Warning:** This is another way a function's behavior can have hidden side effects beyond its `return` value — similar to the problem with global variables above. If a function is supposed to modify a list or dictionary, make that intention clear (e.g. in the function's name, or by also returning the modified object).

**4. It is good practice to place code inside functions:**

Due to the complexity of many programs, we often prefer to place our code inside one or several functions, to increase readability and reusability. This also keeps the namespace of our program smaller, i.e., fewer global variables to keep track of.

We can make function calls inside another function.

````python
def exp(num, exponent):
    res = num**exponent
    return res

def print_result(a, b):
    res = exp(a, b)  # function call to "exp"
    print(res)

print_result(3, 2)
````

To organize our code, it is common to use a single function to run the entire program.

> 📝 **Note:** It is conventional to name this "mother" function `main` — the function that runs the whole program.

The `main` function should make all of the necessary function calls to run the program, and it should return and/or display the outcome.

````python
def exp(num, base=2):
    res = num**base
    return res

def main():  # "main" takes no input

    # Define local variables
    number = 3
    exponent = 2

    # Function call to "exp"
    res = exp(number, exponent)

    # Print result
    print(f"{number}^{exponent} = {res:.2f}")
````

We can now run the entire program by simply calling `main`.

````python
main()
````
