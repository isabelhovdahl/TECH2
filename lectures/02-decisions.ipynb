# 02 - Decisions

Programs are built out of three kinds of control structure:

- **Sequential** - the statements run one after another, from top to bottom. This is the default, and it is all we have used so far.
- **Selection** - the program picks one of several blocks of code to run, depending on a condition. That is this notebook.
- **Iterative** - a block of code is repeated. You have not met loops yet.

So far we have executed our Python programs line-by-line. However, sometimes we want to control the order that instructions (i.e., code) are executed in the program.

This notebook gives an introduction to selection control in Python. In selection control, the program needs to make a *decision* based on a *condition*. These conditions are created by combining **`if` statements** with **boolean expressions**.


## Boolean expressions

These are True/False expressions.

We create these expressions by using the boolean data type: `True` and `False`.

````python
is_raining = True 
is_sunny = False

print(is_raining)
print(type(is_raining))
````

Two things to notice. `True` and `False` are written with a capital first letter, and they are **not** strings.

### Operators

To generate a boolean expression, we must use *operators* to compare values (or expressions) to each other. The output of such an operation will always be equal to one of the boolean values.

The most common operators to create boolean expressions are:
1. Comparison operators
2. Membership operators
3. Logical operators

#### 1. Comparison operators:

Operators that perform the "usual" comparison operations that we are familiar with from math.

|Operator | Description                                                                        | Syntax |
|:---     | :---                                                                               | ---    |
|==       | Equal to: True if both values are equal                                            | x == y |
|>        | Greater than: True if the left value is greater than the right                     | x > y  |
|>=       | Greater than or equal to: True if left value is greater than or equal to the right | x >= y |
|<        | Less than: True if the left value is less than the right                           | x < y  |
|<=       | Less than or equal to: True if left value is less than or equal to the right       | x <= y |
|!=       | Not equal to: True if the values are not equal                                     | x != y |


````python
x = 10
y = 3

print(x == y)
print(x != y)
print(x <= y)
````

> &#x26A0; **Warning:** `=` and `==` are different operators. A single `=` *assigns* a value to a name; a double `==` *asks* whether two values are equal. 

Recall that floats have limited precision - this means we should be careful when comparing them directly.

````python
1 / 3 == 0.333
````

````python
round(1 / 3, 3) == 0.333
````

We can also perform comparison operations on string data. However, for two strings to be equal to each other they must contain the *exact* same sequence of characters.

````python
print("Hello" == "Hello")
print("Hello" == "HELLO")
````

To avoid the issue of upper- and lowercase, we can use the `upper` or `lower` functions when comparing strings.

````python
str_low = "hello"
str_up = "HELLO"

print(str_low.upper() == str_up)
print(str_low == str_up.lower())
````

#### 2. Membership operators:

There are only two membership operators: `in` and `not in`.

We can use these operators to evaluate whether a particular value occurs within a sequence of values.

````python
10 in [40, 20, 10]
````

````python
10 not in [40, 20, 10]
````

````python
valid_grades = ["A", "B", "C", "D", "E", "F"]

print("B" in valid_grades)
print("G" in valid_grades)
print("G" not in valid_grades)
````

As strings are just a sequence of characters, we can use the membership operators to check whether a string occurs within another string.

````python
"Dr." in "Dr. Malone"
````

We can also use the membership operators to check whether a value occurs within a dictionary. However, note that this checks if a given *key* is present in the dictionary.

````python
person = {
    "name": "Alice",
    "age": 30,
    "city": "Bergen"
}
````

````python
print("name" in person)
print("Alice" in person)
````

If we want to check membership of the *values* in a dictionary, we can apply the `values` function to extract a list of the values in the dictionary.

````python
"Alice" in person.values()
````

#### 3. Logical operators:

There are three logical operators: `and`, `or` and `not`.

We can use these logical operators to create more complex boolean expressions by combining two or more boolean expressions into a single expression.

- `A and B` is `True` only when **both** sides are true
- `A or B` is `True` when **at least one** side is true
- `not A` flips true to false and false to true

````python
print(True and True)
print(True and False)
print(False and False)
````

````python
print(True or True)
print(True or False)
print(False or False)
````

````python
print(not True)
print(not False)
````

In practice we never write them on bare `True` and `False`. Instead we write them on comparisons.

````python
age = 25
income = 320_000

print(age >= 18 and income >= 300_000)
print(age >= 30 or income >= 300_000)
print(not age >= 30)
````

### Analyzing strings

In addition to comparing strings using the comparison and membership operators, Python also has string methods that evaluate a string and return a boolean value.

`isdigit` returns `True` if all characters in the string are digits (0-9) and there is at least one character, `False` otherwise.

````python
print("2026".isdigit())
print("abc123".isdigit())
print("-42".isdigit())
print("3.14".isdigit())
````

`islower` returns `True` if all cased characters in the string are lowercase and there is at least one cased character, `False` otherwise.

````python
print("abcdef".islower())
print("Abcdef".islower())
print("abc123".islower())
print("123".islower())
````

Python has many other built-in string methods for analyzing strings beyond the ones demonstrated here. A few of the most useful ones:

- `isalpha()` - `True` if all characters are letters
- `isalnum()` - `True` if all characters are letters or digits
- `isupper()` - `True` if all cased characters are uppercase (the opposite of `islower`)
- `isspace()` - `True` if all characters are whitespace
- `startswith(substring)` / `endswith(substring)` - `True` if the string starts/ends with the given substring
- `istitle()` - `True` if the string follows Title Case (each word starts with an uppercase letter)

You can find the full list in the [Python documentation](https://docs.python.org/3/library/stdtypes.html#string-methods).


## `if` statements

`if` statements allow us to select which part of the code to execute based on a condition.

The statement consists of a header starting with the `if` keyword, followed by a boolean condition and a colon (:), and its associated block of code:
```
if <condition>:

    <statements>

```

Note that the block of code is indented relative to its header, and it will be executed only if the condition is `True`.

````python
score = 100
#score = 70

if score == 100:
    print("Full score!")
````

Note that all code *outside* of the indented code block will still be executed even if the condition is not true.

````python
score = 70 

if score == 100:
    print("Full score!")

print("Always print this.")    
````

We can add alternative instructions (i.e., code) by combining an `if` statement with an `else` statement.

The code inside the `else` statement will only be executed if the condition in the `if` statement is not `True`. Therefore, the `else` statement does not take a condition.

````python
score = 70

if score == 100:
    print("Full score!")
else:
    print("Not full score.")
````

In addition to the comparison operators (e.g. `<=`), we can also use the membership operator `in` in an `if` statement.

````python
passing_grades = ["A", "B", "C", "D", "E"]
grade = "C"

if grade in passing_grades:
    print("You have received a passing grade")
else:
    print("You have not received a passing grade.")
````

We have already seen that we can test for multiple conditions simultaneously using the logical operators:`and` and `or` 

````python
score = 70
#score = 84

if (score < 90) and (score >= 80):
    print("Grade: B")
else:
    print("Grade is not B.")
````

````python
score = 70
#score = 102

if (score < 0) or (score > 100):
    print("Invalid score!")
else:
    print("Valid score.")
````

It is not only `print` statements that we can place inside the `if` statements, but we can perform any type of Python operation.

````python
num = 12

if num > 10:
    num = num / 2
else:
    num = num * 2

print(num)
````

### Nested `if` statements

An `if` statement can contain any kind of code inside its block - including another `if` statement. This is called *nesting*, and it lets us build more complex selection structures.

One common use of nesting is to place an `if` statement inside the `else` block of another `if` statement, so that we only check the second condition when the first one was `False`. Below, we use this to sort a score into a grade.

````python
score = 85

if score >= 90:
    print("Grade: A")
else:
    if score >= 80:
        print("Grade: B")
    else:
        if score >= 70:
            print("Grade: C")
        else:
            print("Grade is not A, B or C.")
````

### `if` statement + `elif` statements

However, nested `if` statements can quickly become very messy. When you have multiple conditions to test, a good alternative is to instead combine the initial `if` statement with one or several `elif` statements.

The program will terminate after the first `if` statement that is evaluated to be `True`. As always, the program should end with a "catch-all" `else` statement.

````python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade is not A, B or C.")
````

Note that there is a key difference between combining an `if` statement with `elif` statements, and instead using multiple separate `if` statements.

With `if` + `elif`, the program stops at the first condition that evaluates to `True` and skips the rest. With separate `if` statements, however, *every* `if` is checked independently - so more than one block can end up executing. Let's see what happens if we rewrite our grade example using separate `if` statements instead of `elif`:

````python
score = 85

if score >= 90:
    print("Grade: A")
if score >= 80:
    print("Grade: B")
if score >= 70:
    print("Grade: C")
else:
    print("Grade is not A, B or C.")
````

Running this prints both `Grade: B` **and** `Grade: C`. This happens because each `if` statement is checked independently: `score >= 80` is `True`, so `Grade: B` is printed - but then the *next*, separate `if` statement is also checked, and `score >= 70` is `True` as well, so `Grade: C` is printed too. The `else` only belongs to the last `if`, so it never gets a chance to run.

With `elif`, the program stops at the first `True` condition - which is exactly the behavior we want when checking mutually exclusive categories like a grade.
