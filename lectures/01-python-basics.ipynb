## Numeric data

Variables containing only digits (with optional sign character and decimal point):
- integers (whole numbers), e.g., 10
- floats (with decimals), e.g., 10.0

> 📝 **Note:** Commas are never used to define floats in Python! Instead, we use the `.` for the decimal point.

We can use the `type` function to get the data type of a variable.

````python
quantity = 10
price = 12.0

print(type(quantity))
print(type(price))
````

Note that evaluating expressions with both integers and floats will always return a float.

````python
eq = 7 + 4.0
eq
````

We can perform all the usual arithmetic operations (e.g., addition, subtraction etc.) on numeric values. Note that exponentiation uses a double multiplication sign.

````python
print(3 ** 2)
````

We can use the `int` and `float` functions to convert a value from one type to the other.

````python
value = 10.0

int(value)
````

However, converting a value does not change the variable. If we want to keep the change, we have to assign the result to a new variable name or overwrite the old one.

````python
# value itself is unchanged
value
````

````python
value = int(value)

value
````

Be careful with `int`: it does not round, it simply drops everything after the decimal point!

````python
int(12.7)
````

Floats also have limited precision in Python (about 16 digits), because computers store numbers in binary rather than decimal. This can cause small roundoff errors in our calculations.

````python
# Should be 0.33333333333333333333...
1/3
````

````python
# Should be exactly 2
1/3 + 1/3 + 1/3 + 1/3 + 1/3 + 1/3
````

We can deal with roundoff errors by rounding to a fixed number of decimals using the `round` function. 

````python
result = 1/3 + 1/3 + 1/3 + 1/3 + 1/3 + 1/3

round(result, 1)
````

> 📝 **Note:** Python uses scientific notation to display very large or very small floats (e.g. `1.5e+16`).

## Strings

In Python, we call text data for *strings*.

A string is any sequence of characters. The characters can be letters, numbers, special characters, spaces etc.

Strings are created by placing the sequence of characters between a pair of single quotes `''` or double quotes `""`.

````python
sentence = "This is how I prefer to write strings"

sentence
````

Strings can contain quotation marks as long as they are different from the ones delimiting the string.

````python
# SyntaxError
quote = "She said "hello" and walked away."
````

````python
# Valid string
quote = 'She said "hello" and walked away.'

quote
````

We can use the `print` function to display a string without the quotation marks.

````python
print(quote)
````

We can use the `str` function to convert a number to a string, and the `int` and `float` functions to convert a string containing only digits back to a number.

````python
num = 10

str(num)
````

````python
str_float = "10.2"

float(str_float)
````

As before, if we want to store the transformation we must store it in a new variable (or overwrite an existing variable).

````python
num_float = float(str_float)

num_float
````

### Working with strings

In general, strings must be contained on one line. However, we can display text across multiple lines by using the newline control character, `\n`.

````python
print("This is a list of numbers: \n1\n2\n3\n4")
````

We can use the `len` function to get the length of a sequence.

````python
word = "Python!"

len(word)
````

#### General sequence methods

Strings are essentially just a sequence of characters, and there are several operations that can be applied to sequences in Python.

**1. String repetition:**

We can use the `*` operator to repeat a string several times.

````python
string1 = "Hello"

print(string1*3)
````

**2. String concatenation:**

We can use the `+` operator to "add" strings together.

````python
print("a" + "b" + "c")
````

````python
string1 = "Hello"
string2 = "world!"

print(string1 + " " + string2)
````

To include a number in a string, we have to first convert it with `str`, otherwise Python complains the it cannot add a string and a number.

````python
temp = 15.4

print("Today's temperature is " + str(temp) + " degrees.")
````

**3. String indexing:**

We can select a specific character from a string by placing its index inside the indexing operator `[]`:
```
string_name[index]
```

> 📝 **Note:** Python uses zero-based indexing. The first item in a sequence is at index 0, not 1.

````python
greeting = "Hello!"

greeting[0]
````

````python
greeting[1]
````

This means the last index in a string is its length minus 1. Recall that we can use the `len` function to get the length of a sequence.

````python
len(greeting)
````

````python
# IndexError
string1[6]
````

````python
greeting[len(greeting) - 1]
````

Counting to the end of a long string is tedious, so Python lets us use negative indices, which are interpreted as positions counted from the end.

````python
print(greeting[-1])
print(greeting[-2])
````

**4. String slicing:**

We can also use the index operator `[]` to extract substrings from a string, which is known as slicing:
```
string_name[start:stop]
```

When slicing, we extract the character from `start` up to but **not including** `stop`.

````python
sentence = "This is an intro class to Python"
````

````python
# Extract first word
sentence[0:4]
````

We can leave out either index. If we omit the start, Python slices from the beginning of the sequence; if we omit the stop, it slices to the end.

````python
# From the beginning
sentence[:5]
````

````python
# To the end
sentence[26:]
````

Negative indices work in slices too, which is the easiest way to take something off the end.

````python
# The last six characters
sentence[-6:]
````

## Lists

Lists in Python are very similar to our everyday concept of lists. In the real world, we read off, add, and cross off items in lists, and we can do the same with lists in Python.

Lists are one of the most common ways to store data in Python, and they are very useful when we want to store a collection of items in a single variable.

We create lists by placing a comma-separated sequence of items within a pair of square brackets:
```
lst_name = [item1, item2, item3, ...]

```

> &#x26A0; **Warning:** Never name a list `list`, since this is already the name of a built-in Python function.

````python
fruits = ["banana", "apple", "cherry"]

fruits
````

````python
print(type(fruits))
print(len(fruits))
````

A list can contain any Python data type, and the items do not all have to be of the same type.

````python
mixed_lst = ["banana", 2, 42.5, 10*2]

mixed_lst
````

#### Indexing and slicing

A list is a sequence, so the index operator works exactly as it did for strings. Same zero-based indexing, same negative indices, same slicing rules.

````python
print(mixed_lst[0])
print(mixed_lst[-1])
print(mixed_lst[1:3])
````

> 📝 **Note:** Slicing a list always returns a *list*, even when the slice contains a single item.

````python
mixed_lst[0:1]
````

#### Operations

Unlike numbers and strings, lists are *mutable*: we can change a list after it has been created without assigning the result to a new variable.

We can add a new item to the end of the list using `append`.

````python
fruits
````

````python
fruits.append("pear")
````

````python
fruits
````

Instead of appending, we can use the `+` operator to combine several lists together into a single list.

````python
vegetables = ["potato", "carrot", "broccoli", "onion"]
````

````python
fruits + vegetables
````

However, because this creates a new list rather than modifying an existing one, we have to store the result.

````python
new_lst = fruits + vegetables

new_lst
````

When a list contains only numbers, there are some special operations that we can perform on the list.

````python
num_lst = [1, 2, 3, 4, 5, 6, 7]

print(sum(num_lst))
print(min(num_lst))
print(max(num_lst))
````

> 📝 **Note:** Python also has *tuples*, created with parentheses `()` instead of square brackets, e.g. `("banana", "apple", "cherry")`. Tuples work like lists but are *immutable* — they cannot be changed after creation. 

#### Nested list

A list can contain other lists. This is known as a nested list, and it is how we represent a table before we have anything better.

````python
foods = [fruits, vegetables]

foods
````

````python
# The first list
foods[0]
````

````python
# The second item in first list
foods[0][1]
````

## Dictionaries

A dictionary is another data type used to store several items in a single variable. Dictionaries in Python are like dictionaries in the real world: we look up words (the **key**) and read off the definition (the **value**).

Unlike lists, each item in a dictionary is associated with a key rather than a position. We create a dictionary by placing comma-separated *key-value* pairs inside curly brackets `{}`. Each pair starts with the key, followed by a colon `:` and then the value:

```
dict_name = {key1: value1, key2: value2, key3: value3, ...}

```

> &#x26A0; **Warning:** Never name a dictionary `dict`, since this is already the name of a built-in Python function.

To make a dictionary readable, we often write each key-value pair on its own line.

````python
student = {
    "name"      : "Anne Smith",
    "student_no": "s1234",
    "course"    : "TECH2",
    "score"     :  82
}

student
````

Like a list, a dictionary has a length, which is the number of key-value pairs it contains.

````python
len(student)
````

But unlike a list, a dictionary has no positions, so we cannot select an item using a numeric index.

````python
# KeyError
student[0]
````

Instead, we use the key to look up a value. We can use the `keys` method to see which keys are available, and the `values` method to see the values.

````python
student.keys()
````

````python
student.values()
````

````python
print(student["name"])
print(student["score"])
````

Like lists, dictionaries are mutable. We can update the value of an existing key, and we can add entirely new key-value pairs.

````python
# Update the value of an existing key
student["score"] = 92

# Add a new key-value pair
student["university"] = "NHH"

student
````

The values in a dictionary can be of any type, including lists. This allows us to store several observations under a single key.

````python
student["score"] = [92, 87, 82]

student
````

And we can then index into that list, exactly as we did with nested lists.

````python
# The second score
student["score"][1]
````
