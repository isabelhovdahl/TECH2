## Numeric data

Variables containing only digits (with optional sign character and decimal point):
- integers (whole numbers), e.g., 10
- floats (with decimals), e.g., 10.0

> 📝 **Note:** Commas are never used to define floats in Python! Instead, we use the `.` for the decimal point.

````python
x = 12.5
````

Note that evaluating expressions with both integers and floats will always return a float.

````python
eq = 7 + 4.0
eq
````

We can use the `type` function to get the data type of a variable.

````python
type(eq)
````

We can use the `int` and `float` functions to convert the data type of numeric data.

````python
num = 10.0
num
````

````python
int(num)
````

However, if we want to actually store the change in the data type, then we have to explicitly re-assign the new value to the previous variable (or store it in a new variable).

````python
num
````

````python
num = int(num)
num
````

````python
num = float(num)
num
````

Note that `int` will simply drop everything after the decimal point.

````python
int(12.7)
````

> 📝 **Note:** Python uses scientific notation to display very large or very small floats (e.g. `1.5e+16`). Floats also have limited precision in Python (about 16 digits), because computers store numbers in binary rather than decimal. This can cause small roundoff errors in our calculations.

````python
# Should be equal to 2
1/3 + 1/3 + 1/3 + 1/3 + 1/3 + 1/3
````

We can deal with roundoff errors by rounding to a fixed number of decimals using the `round` function.

````python
result = 1/3 + 1/3 + 1/3 + 1/3 + 1/3 + 1/3

round(result, 1)
````

## Strings

In Python, we call text data for *strings*.

A string is any sequence of characters. The characters can be letters, numbers, special characters, spaces etc.

Strings are created by placing the sequence of characters between a pair of single quotes `''` or double quotes `""`.

````python
sentence = "This is how I prefer to write strings"
sentence
````

Strings can contain quotation marks as long as different quotation marks are used to delimit the string. This is one reason why we default to double quotes: it lets us write text containing apostrophes without extra escaping.

````python
# SyntaxError
sentence = 'We're having fun with Python today!'
````

````python
# Valid string
sentence = "We're having fun with Python today!"
sentence
````

Note that we can use the `print` function to display a string without the quotation marks.

````python
print(sentence)
````

We can use the `str` function to convert a number to a string.

````python
num = 10
str(num)
````

And we can also use the `int` and `float` functions to convert a string (with only digits) to a number.

````python
str_int = "10"
int(str_int)
````

````python
str_float = "10.2"
float(str_float)
````

As before, if we want to store the transformation we must store it in a new variable (or overwrite an existing variable).

````python
num_float = float(str_float)
num_float
````

### Working with strings

In general, strings must be contained on one line. However, we can display text across multiple lines by using the newline control character, `\n`.

````python
print("This is a list of numbers: \n1\n2\n3\n4")
````

We can use the `len` function to get the length of a sequence.

````python
word = "Python!"
len(word)
````

#### General sequence methods

Strings are essentially just a sequence of characters, and there are several operations that can be applied to sequences in Python.

**1. String repetition:**

We can use the `*` operator to repeat a string several times.

````python
string1 = "Hello"

print(string1*3)
````

**2. String concatenation:**

We can use the `+` operator to "add" strings together.

````python
print("a" + "b" + "c")
````

````python
string1 = "Hello"
string2 = "world!"

print(string1 + " " + string2)
````

Recall that we can add a number to a string by using the `str` function to convert the number to a string.

````python
temp = 15.4

print("Today's temperature is " + str(temp) + " degrees.")
````

**3. String indexing:**

We can select a specific character from a string by placing its index inside the indexing operator `[]`:
```
string_name[index]
```

> 📝 **Note:** Python uses zero-based indexing.

````python
string1 = "Hello!"
print(string1)
````

````python
string1[1]
````

````python
string1[0]
````

This means the last index in a string is its length minus 1. Recall that we can use the `len` function to get the length of a sequence.

````python
len(string1)
````

````python
# IndexError
string1[6]
````

````python
string1[len(string1)-1]
````

**4. String slicing:**

We can also use the index operator `[]` to extract substrings from a string, which is known as slicing:
```
string_name[index1:index2]
```

When slicing, we extract the character from `index1` up to (but not including) `index2`.

````python
sentence = "This is an intro class to Python"
````

````python
# Extract first character
sentence[0]
````

````python
# Alternative
sentence[0:1]
````

````python
# Extract first word
sentence[0:4]
````

By default, the index operator will extract all characters starting from the first to the last index.

````python
# Extract first word + space (omit start index)
sentence[:5]
````

````python
# Extract last word (omit end index)
sentence[26:]
````

A negative index is interpreted as the number of characters from the *end* of the string.

````python
sentence[-6:]
````

## Lists

Lists in Python are very similar to our everyday concept of lists. In the real world, we read off, add, and cross off items in lists, and we can do the same with lists in Python.

Lists are one of the most common ways to store data in Python, and they are very useful when we want to store a collection of items in a single variable.

We create lists by placing a comma-separated sequence of items within a pair of square brackets `[]`:
```
lst_name = [item1, item2, item3, ...]

```

````python
fruits = ["banana", "apple", "cherry"]

fruits
````

````python
type(fruits)
````

````python
len(fruits)
````

Lists in Python can contain all types of data, e.g., strings, integers, floats, arithmetic operations etc.

````python
mixed_lst = ["banana", 2, 42.5, 10*2]

mixed_lst
````

#### Indexing and slicing

We select items from a list the same way that we selected characters from a string by using the index operator `[]`.

Recall that Python follows *zero-based indexing*.

````python
mixed_lst[0]
````

````python
mixed_lst[3]
````

````python
mixed_lst[2:]
````

As before, negative numbers are interpreted as the number of items from the *end* of the list.

````python
mixed_lst[-3:-1]
````

#### Operations

Unlike numbers and strings, lists are *mutable*, meaning that we can make changes to a list without having to assign it to a new variable (or overwrite a previous variable).

We can add a new item to the end of the list using `append`.

````python
fruits
````

````python
fruits.append("pear")
````

````python
fruits
````

> 📝 **Note:** Python also has *tuples*, created with parentheses `()` instead of square brackets, e.g. `("banana", "apple", "cherry")`. Tuples work like lists but are *immutable* — they cannot be changed after creation. More on this later.

Instead of appending, we can use the `+` operator to combine several lists together into a single list.

````python
vegetables = ["potato", "carrot", "broccoli", "onion"]
````

````python
fruits + vegetables
````

However, to store the appended list we need to store it in a new variable (or overwrite a previous variable).

````python
new_lst = fruits + vegetables

new_lst
````

When a list consists of only numbers, there are some special operations that we can perform on the list, such as `sum`, `min` and `max`.

````python
num_lst = [1, 2, 3, 4, 5, 6, 7]

num_lst
````

````python
sum(num_lst)
````

````python
min(num_lst)
````

````python
max(num_lst)
````

> &#x26A0; **Warning:** Never name a list `list`, since this is already the name of a built-in Python function.

#### Nested list

A list can also contain other lists. This is known as a *nested* list.

````python
foods = [fruits, vegetables]

foods
````

````python
# First list
foods[0]
````

````python
# Second list
foods[1]
````

````python
# Second item in first list
foods[0][1]
````

## Dictionaries

A dictionary is another data type used to store multiple items in a single variable. Dictionaries in Python are like dictionaries in the real world: we look up words (keys) and read off the definition (value).

Unlike lists, each item in a dictionary is associated with a *key* instead of an index. We create dictionaries by placing a comma-separated sequence of *key-value* pairs within curly brackets `{}`. Each key-value pair starts with the key, followed by a colon `:` and then the value associated with that key:

```
dict_name = {key1: value1, key2: value2, key3: value3, ...}

```

To make the dictionary more readable to humans, we often write each key-value pair on its own line.

````python
student = {
    "name" : "Anne Smith",
    "student_no" : "s1234",
    "course" : "TECH2",
    "score" :  82
}

student
````

Like lists, dictionaries are also a sequence of items.

````python
len(student)
````

But unlike lists, dictionaries have no index, so we cannot select an item from a dictionary using the indexing operator.

````python
# KeyError
student[0]
````

Instead, we have to use the keys to "look up" a value in a dictionary.

We can use the `keys` function to display all of the keys in a dictionary.

````python
student.keys()
````

````python
student["name"]
````

````python
student["score"]
````

Like lists, dictionaries are also mutable. This means that we can update the value of an existing key and add new key-value pairs.

````python
# Update value of existing key
student["score"] = 92

student
````

````python
# Add new key-value pair
student["university"] = "NHH"

student
````

Values in a dictionary can be of any Python data type, including lists.

````python
student["score"] = [92, 87, 82]

student
````
