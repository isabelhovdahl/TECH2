# Notebook & Markdown Demo

A notebook mixes two kinds of cells:
- **Markdown cells** - formatted text, for explaining what's going on
- **Code cells** - actual Python code you can run

This is different from a script, where everything is just code from top to bottom. In a notebook, you can mix explanations and code however you like - which makes notebooks great for documenting and walking through your work step by step.

## Running cells

You can run a cell by selecting it and pressing `Shift + Enter`, or by clicking the ▶ button that appears to the left of the cell.

What happens when you run a cell depends on its type.

### Code cells

If a cell is a **Code** cell, running it executes the Python code inside. Notice that a notebook automatically displays the result of the *last* line - unlike in a script, we don't need `print()` here.

```python
2 + 2
```

### Markdown cells

If a cell is a **Markdown** cell, running it displays the text as nicely formatted output instead of executing it as code. This very cell is a markdown cell!

To see the raw markdown text behind any formatted cell (including this one), just double-click it. Running it again (`Shift + Enter`) turns it back into formatted text.

### Adding and deleting cells

Hover just below a cell to see **+ Code** and **+ Markdown** buttons for adding a new cell. To delete a cell, click the trash icon in the cell's toolbar (visible when you hover over the cell).

## Markdown formatting

Markdown cells support simple formatting. You'll see plenty more examples in the lecture notebooks - here are the three you'll use most.

**Titles**, using one or more `#`:

# Title
## Subtitle
### Smaller heading

**Bold text**, using double asterisks:

This is **bold text**, and this is regular text.

**Lists**, using `-` for bullet points or numbers for an ordered list:

- item 1
- item 2
- item 3

1. first step
2. second step
3. third step

## Variables

In programming, we often store the inputs and outputs of our code in variables.

A variable is a named storage location in the computer that associates a *name* (identifier) with a specific *value*.

Variables are created by assigning an initial value to an identifier using the equality sign (=).

```python
num = 10
num
```

Python is a *dynamically-typed* language, which means that we can change the value of a variable throughout a program. In other words, we can overwrite variables.

```python
num = num + 10
num
```

Although variable names must be unique, we can assign the same value to multiple variables.

```python
var1 = 10
var2 = 10

print(var1)
print(var2)
```

In general, we should try to give our variables intuitive names. However, there are a few rules that we must follow when naming variables in Python. A variable name cannot:
- start with a number
- contain spaces
- contain special characters, e.g. !, ', #, @ etc.

```python
# SyntaxError: variable name contains spaces
six pack = 6
```

```python
# Instead, we can use underscores in names (snake case)
six_pack = 6
```

```python
# Or capitalize letters (camel case)
sixPack = 6
```

That's the pattern to get used to: use markdown cells like the ones above to explain *what* your code does and *why*, right next to the code itself. You'll see this same style throughout the lecture notebooks.
